# Parte 3: Limpieza de datos

Se trabaja sobre el subconjunto de Miami (`df_miami`). Pasos: **duplicados → errores (valores imposibles) → nulos → normalización de formatos**, con verificación final.

## 3.1 Duplicados

```python
print("Filas duplicadas completas:", df_miami.duplicated().sum())
print("IDs de persona duplicados:", df_miami["id_persona"].duplicated().sum())

df_limpio = df_miami.drop_duplicates().copy()
```

## 3.2 Errores: valores imposibles → se convierten a nulo para imputarlos después

```python
err_edad = ((df_limpio["edad"] < 0) | (df_limpio["edad"] > 110)).sum()
err_frec = (df_limpio["frecuencia_visita"] < 0).sum()
print("Edades inválidas (<0 o >110):", err_edad)
print("Frecuencias de visita negativas:", err_frec)

df_limpio.loc[(df_limpio["edad"] < 0) | (df_limpio["edad"] > 110), "edad"] = pd.NA
df_limpio.loc[df_limpio["frecuencia_visita"] < 0, "frecuencia_visita"] = pd.NA

# Chequeo de consistencia: ¿hay clientes con frecuencia 0 pero gasto > 0, o al revés?
print("frecuencia=0 con gasto>0:", ((df_limpio["frecuencia_visita"] == 0) & (df_limpio["promedio_gasto_comida"] > 0)).sum())
print("frecuencia>0 con gasto=0:", ((df_limpio["frecuencia_visita"] > 0) & (df_limpio["promedio_gasto_comida"] == 0)).sum())
```

## 3.3 Nulos: diagnóstico e imputación

```python
print("Nulos antes de imputar:")
print(df_limpio.isna().sum()[lambda s: s > 0])

# Numéricas → mediana (robusta frente a outliers)
for col in ["edad", "frecuencia_visita", "promedio_gasto_comida"]:
    mediana = df_limpio[col].median()
    df_limpio[col] = df_limpio[col].fillna(mediana)
    print(f"{col}: imputado con mediana = {mediana}")

# Categóricas / contacto → etiqueta explícita (no se inventan datos personales)
df_limpio["preferencias_alimenticias"] = df_limpio["preferencias_alimenticias"].fillna("No especificado")
df_limpio["telefono_contacto"] = df_limpio["telefono_contacto"].fillna("Sin dato")
df_limpio["correo_electronico"] = df_limpio["correo_electronico"].fillna("Sin dato")

# Tipos finales coherentes
df_limpio["edad"] = df_limpio["edad"].astype(int)
df_limpio["frecuencia_visita"] = df_limpio["frecuencia_visita"].astype(int)
```

## 3.4 Normalización del teléfono al formato XXX-XXX-XXXX (conservando extensiones)

```python
import re

def normalizar_tel(t):
    if t == "Sin dato":
        return t
    principal, _, ext = str(t).lower().partition("x")
    d = re.sub(r"\D", "", principal)
    if len(d) == 11 and d.startswith("1"):      # prefijo +1
        d = d[1:]
    elif len(d) == 13 and d.startswith("001"):  # prefijo 001
        d = d[3:]
    if len(d) == 10:
        tel = f"{d[0:3]}-{d[3:6]}-{d[6:10]}"
        return f"{tel} ext.{ext}" if ext else tel
    return t

df_limpio["telefono_contacto"] = df_limpio["telefono_contacto"].map(normalizar_tel)

# Verificación: todos los teléfonos con dato cumplen el formato estándar
con_dato = df_limpio.loc[df_limpio["telefono_contacto"] != "Sin dato", "telefono_contacto"]
no_norm = (~con_dato.str.match(r"^\d{3}-\d{3}-\d{4}( ext\.\d+)?$")).sum()
print("Teléfonos sin normalizar:", no_norm)
con_dato.head()
```

## 3.5 Verificación final y guardado

```python
print("Nulos restantes:", df_limpio.isna().sum().sum())
print("Dimensiones finales:", df_limpio.shape)
print(df_limpio[["edad", "frecuencia_visita", "promedio_gasto_comida"]].describe())

df_limpio.to_csv("../../Datasets/clientes_miami_limpio.csv", index=False, encoding="utf-8")
print("\nGuardado: clientes_miami_limpio.csv")
```

## Conclusiones de la limpieza

| Problema | Hallazgo | Tratamiento |
|---|---|---|
| Duplicados | 0 filas y 0 IDs repetidos | Nada que eliminar (se valida igualmente) |
| Edades imposibles | 28 (valores < 0 o > 110) | Convertidas a nulo e imputadas con la mediana (49) |
| Frecuencias negativas | 173 | Convertidas a nulo e imputadas con la mediana (5) |
| Nulos en `edad` | 36 en total (8 originales + 28 por error) | Mediana (49) |
| Nulos en `promedio_gasto_comida` | 16 | Mediana ($31,6) |
| Nulos en `preferencias_alimenticias` | 146 | Etiqueta "No especificado" |
| Nulos en teléfono / correo | 1.596 / 1.589 (~50%) | Etiqueta "Sin dato" — no se inventan datos personales |
| Formatos de teléfono inconsistentes | 5+ estilos distintos (paréntesis, puntos, prefijos +1/001, extensiones) | Normalizados a `XXX-XXX-XXXX` (+ `ext.` si aplica); 0 sin normalizar |

**Decisiones clave:**
- Se usó la **mediana** (no la media) para imputar numéricas, porque es robusta frente a la distribución sesgada del gasto y a los outliers.
- Los valores imposibles se trataron como errores de carga: no se puede "adivinar" el valor real, así que se imputaron igual que los nulos.
- No se eliminó ninguna fila: el dataset limpio conserva los **3.186 clientes**, ahora sin nulos y con rangos válidos (edad 18–80, frecuencia 0–10).
- Resultado guardado en `clientes_miami_limpio.csv`, listo para la etapa de análisis.

# Parte 4: Tablas de distribución

Tablas básicas para entender cómo se distribuyen las variables del dataset limpio de clientes de Miami: género, edad, gasto, frecuencia de visita, estrato, preferencias alimenticias, consumo de licor, membresía y tipo de pago.

## 4.1 Tablas de frecuencia de variables categóricas (cantidad y %)

```python
def tabla_frecuencia(serie, nombre):
    t = serie.value_counts().to_frame("clientes")
    t["%"] = (t["clientes"] / len(serie) * 100).round(1)
    print(f"=== {nombre} ===")
    print(t, "\n")

tabla_frecuencia(df_limpio["genero"], "Género")
tabla_frecuencia(df_limpio["estrato_socioeconomico"], "Estrato socioeconómico")
tabla_frecuencia(df_limpio["preferencias_alimenticias"], "Preferencias alimenticias")
tabla_frecuencia(df_limpio["membresia_premium"], "Membresía premium")
tabla_frecuencia(df_limpio["tipo_de_pago_mas_usado"], "Tipo de pago más usado")
tabla_frecuencia(df_limpio["consume_licor"], "Consume licor")
tabla_frecuencia(df_limpio["ocio"], "Sale por ocio")
```

## 4.2 Variables numéricas agrupadas en rangos

```python
df_limpio["rango_edad"] = pd.cut(df_limpio["edad"], bins=[17, 30, 45, 60, 80],
                                 labels=["18-30", "31-45", "46-60", "61-80"])
tabla_frecuencia(df_limpio["rango_edad"], "Rango de edad")

df_limpio["rango_gasto"] = pd.cut(df_limpio["promedio_gasto_comida"], bins=[-0.01, 20, 40, 70, 130],
                                  labels=["$0-20", "$20-40", "$40-70", "$70+"])
tabla_frecuencia(df_limpio["rango_gasto"], "Rango de gasto por comida")

tabla_frecuencia(df_limpio["frecuencia_visita"], "Frecuencia de visita (veces/mes)")
```

## 4.3 Tablas cruzadas: cómo se relacionan las variables entre sí

```python
print("=== Gasto por comida según estrato ===")
print(df_limpio.groupby("estrato_socioeconomico")["promedio_gasto_comida"].agg(["mean", "median", "count"]).round(2))

print("\n=== Membresía premium según género (%) ===")
print((pd.crosstab(df_limpio["genero"], df_limpio["membresia_premium"], normalize="index") * 100).round(1))

print("\n=== Gasto medio según rango de edad ===")
print(df_limpio.groupby("rango_edad", observed=True)["promedio_gasto_comida"].agg(["mean", "count"]).round(2))
```

## Conclusiones de las tablas de distribución

- **Género:** prácticamente balanceado — 50,7% masculino, 49,3% femenino. La membresía premium tampoco varía por género (53,5% vs 52,8%).
- **Edad:** el grupo más numeroso es el de **61-80 años (30,7%)**; los jóvenes de 18-30 son solo el 18,8%. El gasto medio es parejo entre edades (~$38-41), así que la edad no discrimina el gasto.
- **Preferencias alimenticias:** domina **Mariscos (25,7%)**, seguido de Vegetariano (22,8%) — coherente con una ciudad costera.
- **Consumo de licor:** el **60%** de los clientes consume licor.
- **Gasto por comida:** la franja más común es **$20-40 (37%)**; solo el 12% gasta más de $70.
- **Frecuencia de visita:** moda en **5 visitas/mes (27,8%)**; un 3,2% no visita nunca (frecuencia 0).
- **Estrato es la variable que más separa el gasto:** Muy Alto gasta en promedio **$68,6**, Alto $40,1, Medio $25,0 y Bajo $13,3 — una relación casi perfectamente escalonada.
- **Pago:** efectivo (38,3%) y tarjeta (32,3%) dominan; criptomoneda es marginal (2%).
